Question 2: Does tenure affect churn?

```python
bins = [0, 12, 24, 48, 72]
labels = ["0-12", "13-24", "25-48", "49-72"]

df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=bins,
    labels=labels
)

pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100
```

"Tenure has a strong association with customer churn. The churn rate is highest among customers with 0–12 months of tenure (47.7%) and steadily decreases to 9.5% among customers with 49–72 months of tenure. This suggests that customer retention efforts should particularly focus on the early stages of the customer lifecycle."

Customer Segmentation

primary high-risk segment -> New + Month-to-month + High monthly charges

```python
df["HighCharges"] = df["MonthlyCharges"] > df["MonthlyCharges"].median()
df["CustomerSegment"] = "Other"

df.loc[
    (df["TenureGroup"] == "0-12") &
    (df["Contract"] == "Month-to-month") &
    (df["HighCharges"]),
    "CustomerSegment"]="New + Month-to-month + High charges"
```

secondary high-risk segment:

Existing + Month-to-month + High monthly charges

```python
df.loc[
    (df["TenureGroup"].isin(["13-24", "25-48","48-72"])) &
    (df["Contract"] == "Month-to-month") &
    (df["HighCharges"]),
    "CustomerSegment"]="Existing + Month-to-month + High charges"
```

Low-risk comparison segment:

Long-tenure + Long-term contract

```python
df.loc[
    (df["TenureGroup"] == "49-72") &
    (df["Contract"].isin(["One year", "Two year"])),
    "CustomerSegment"
] = "Long-tenure + Long-term contract"
```

```python
pd.crosstab(
    df["CustomerSegment"],
    df["Churn"],
    normalize="index"
) * 100
```

Conclusion – Customer Risk Segmentation

The customer segmentation analysis reveals a clear high-risk customer segment. Customers who are new (0–12 months), on month-to-month contracts, and have high monthly charges have a 69.95% churn rate, making them the most vulnerable group identified in the analysis.

The existing month-to-month customers with high charges also show elevated churn at 45.10%, indicating that the combination of flexible contracts and higher charges is associated with increased churn even among customers with longer tenure.

In contrast, long-tenure customers on long-term contracts have a much lower churn rate of only 6.54%, suggesting that longer tenure and contractual commitment are associated with greater customer retention.

Overall, the analysis suggests that the company should prioritize newly acquired, high-charge, month-to-month customers for targeted retention efforts, while using long-tenure, long-term customers as a relatively stable customer segment for comparison. These findings indicate associations with churn rather than causation and should be considered alongside the size and revenue contribution of each segment when deciding where to focus retention resources.

Is there a statistically significant association between contract type and churn?

We can use a Chi-square test of independence.

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-square:", chi2)
print("p-value:", p)
```

- Here p-value is > 0.05
- Therefore we cannot reject the null hypothesis
- There is a statistically significant association between contract type and churn
